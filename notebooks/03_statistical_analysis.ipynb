**Objective:**
Compute Pearson and Spearman rank correlation matrices across maternal health indicators and socioeconomic determinants (female literacy, poverty headcount, rurality).


> ⚠️ **Synthetic Demonstration Dataset Notice**
>
> The dataset used in this notebook (datasets/demo/india_maternal_health_district_data.csv) is a **synthetic demonstration dataset** generated programmatically via Gaussian noise around NFHS-5 state-level published averages. It does **not** contain actual district-level survey observations.
>
> All analytical results, metrics, and visualizations produced in this notebook are **demonstration outputs only** and should not be interpreted as reflecting real maternal healthcare conditions across India.
>
> See docs/data_provenance.md for complete provenance documentation.

```python
import pandas as pd
from scipy import stats

df = pd.read_csv('../datasets/demo/india_maternal_health_district_data.csv')
num_cols = [
    'anc4_or_more_pct', 'institutional_delivery_pct', 'skilled_birth_attendance_pct',
    'postnatal_care_pct', 'ifa_100_days_pct', 'female_literacy_rate_pct', 'below_poverty_line_pct'
]
corr_matrix = df[num_cols].corr()
corr_matrix.round(3)
```

```python
# Hypothesis Testing: Does female literacy correlate significantly with institutional births?
r, p_val = stats.pearsonr(df['female_literacy_rate_pct'], df['institutional_delivery_pct'])
print(f"Pearson r = {r:.3f}, p-value = {p_val:.4e}")
assert p_val < 0.05, "Statistically significant correlation confirmed."
```

### Key Findings:
Strong positive correlation confirms socioeconomic enablement is closely tied to formal healthcare access.
