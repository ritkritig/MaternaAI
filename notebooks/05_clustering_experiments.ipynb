### MaternaAI Research Platform

**Objective:**
Apply K-Means clustering to partition districts into homogeneous healthcare accessibility profiles without supervised bias. Validate cluster count using silhouette analysis.


> ⚠️ **Synthetic Demonstration Dataset Notice**
>
> The dataset used in this notebook (datasets/demo/india_maternal_health_district_data.csv) is a **synthetic demonstration dataset** generated programmatically via Gaussian noise around NFHS-5 state-level published averages. It does **not** contain actual district-level survey observations.
>
> All analytical results, metrics, and visualizations produced in this notebook are **demonstration outputs only** and should not be interpreted as reflecting real maternal healthcare conditions across India.
>
> See docs/data_provenance.md for complete provenance documentation.

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

df = pd.read_csv('../datasets/demo/india_maternal_health_district_data.csv')
features = ['anc4_or_more_pct', 'institutional_delivery_pct', 'skilled_birth_attendance_pct', 'postnatal_care_pct', 'ifa_100_days_pct']
X = StandardScaler().fit_transform(df[features])

# Compare silhouette scores for k=2..6
for k in range(2, 7):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X)
    print(f"k={k}: Silhouette Score = {silhouette_score(X, labels):.4f}")
```

```python
# Fit k=3 model & inspect profile means
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10).fit(X)
df['cluster'] = kmeans.labels_
df.groupby('cluster')[features].mean().round(1)
```
