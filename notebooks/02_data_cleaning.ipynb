# Notebook 02: Data Cleaning, Imputation & Outlier Validation
### MaternaAI Research Platform

**Objective:**
Clean raw district observations, detect and treat outliers using Interquartile Range (IQR), evaluate missingness, and produce a standardized cleaned dataset ready for modeling.


> ⚠️ **Synthetic Demonstration Dataset Notice**
>
> The dataset used in this notebook (datasets/demo/india_maternal_health_district_data.csv) is a **synthetic demonstration dataset** generated programmatically via Gaussian noise around NFHS-5 state-level published averages. It does **not** contain actual district-level survey observations.
>
> All analytical results, metrics, and visualizations produced in this notebook are **demonstration outputs only** and should not be interpreted as reflecting real maternal healthcare conditions across India.
>
> See docs/data_provenance.md for complete provenance documentation.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../datasets/demo/india_maternal_health_district_data.csv')

# Missingness check
print("Missing values per feature:")
print(df.isnull().sum())
```

```python
# Outlier analysis using IQR
for col in ['anc4_or_more_pct', 'institutional_delivery_pct', 'maternal_mortality_ratio_proxy']:
    q25 = df[col].quantile(0.25)
    q75 = df[col].quantile(0.75)
    iqr = q75 - q25
    lower = q25 - 1.5 * iqr
    upper = q75 + 1.5 * iqr
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col}: {len(outliers)} outliers detected outside [{lower:.1f}, {upper:.1f}]")
```
