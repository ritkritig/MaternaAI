# Notebook 06: Predictive Modeling (Random Forest Regressor vs Linear Baseline)
### MaternaAI Research Platform

**Objective:**
Train a Random Forest Regressor to estimate district-level institutional delivery coverage from socioeconomic and health infrastructure determinants.


> ⚠️ **Synthetic Demonstration Dataset Notice**
>
> The dataset used in this notebook (datasets/demo/india_maternal_health_district_data.csv) is a **synthetic demonstration dataset** generated programmatically via Gaussian noise around NFHS-5 state-level published averages. It does **not** contain actual district-level survey observations.
>
> All analytical results, metrics, and visualizations produced in this notebook are **demonstration outputs only** and should not be interpreted as reflecting real maternal healthcare conditions across India.
>
> See docs/data_provenance.md for complete provenance documentation.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

df = pd.read_csv('../datasets/demo/india_maternal_health_district_data.csv')
features = ['anc4_or_more_pct', 'female_literacy_rate_pct', 'rural_population_pct', 'below_poverty_line_pct', 'phc_chc_density_per_100k']
target = 'institutional_delivery_pct'

X_train, X_test, y_train, y_test = train_test_split(df[features], df[target], test_size=0.2, random_state=42)

rf = RandomForestRegressor(n_estimators=100, random_state=42).fit(X_train, y_train)
lr = LinearRegression().fit(X_train, y_train)

print(f"Random Forest Test R²: {r2_score(y_test, rf.predict(X_test)):.3f}, MAE: {mean_absolute_error(y_test, rf.predict(X_test)):.2f}%")
print(f"Baseline Linear Test R²: {r2_score(y_test, lr.predict(X_test)):.3f}, MAE: {mean_absolute_error(y_test, lr.predict(X_test)):.2f}%")
```
