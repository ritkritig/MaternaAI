# Notebook 07: Rigorous Cross-Validation & Feature Explainability
### MaternaAI Research Platform

**Objective:**
Perform 5-fold cross-validation, inspect residual distributions, and extract feature importances to explain which determinants drive regional service predictions.


> ⚠️ **Synthetic Demonstration Dataset Notice**
>
> The dataset used in this notebook (datasets/demo/india_maternal_health_district_data.csv) is a **synthetic demonstration dataset** generated programmatically via Gaussian noise around NFHS-5 state-level published averages. It does **not** contain actual district-level survey observations.
>
> All analytical results, metrics, and visualizations produced in this notebook are **demonstration outputs only** and should not be interpreted as reflecting real maternal healthcare conditions across India.
>
> See docs/data_provenance.md for complete provenance documentation.

```python
import pandas as pd
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestRegressor

df = pd.read_csv('../datasets/demo/india_maternal_health_district_data.csv')
features = ['anc4_or_more_pct', 'female_literacy_rate_pct', 'rural_population_pct', 'below_poverty_line_pct', 'phc_chc_density_per_100k']
target = 'institutional_delivery_pct'

rf = RandomForestRegressor(n_estimators=100, random_state=42)
scores = cross_val_score(rf, df[features], df[target], cv=5, scoring='r2')
print(f"5-Fold Cross Validation R²: Mean = {scores.mean():.3f}, Std = {scores.std():.3f}")

rf.fit(df[features], df[target])
importances = pd.Series(rf.feature_importances_, index=features).sort_values(ascending=False)
print("
Feature Importances:")
print(importances.round(4))
```

### Ethical & Methodological Disclaimer:
High feature importance reflects co-variation and predictive utility within survey strata, not causal health guarantees.
